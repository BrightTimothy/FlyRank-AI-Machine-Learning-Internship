## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/BrightTimothy/FlyRank-AI-Machine-Learning-Internship/main/data/raw/content_refresh_anonymized.csv")

# Reason code: low_ctr_visible_page
# impressions_90d >= 500, avg_position between 1-20, ctr < 0.5
low_ctr = df[
    (df["impressions_90d"] >= 500) &
    (df["avg_position"] > 0) & (df["avg_position"] <= 20) &
    (df["ctr"] < 0.5)
]
print("Low-CTR visible pages:", len(low_ctr), f"({len(low_ctr)/len(df)*100:.1f}% of dataset)")

# Reason code: low_engagement_visible_page
# sessions_90d >= 30, engagement_rate or scroll_rate below 30
low_engagement = df[
    (df["sessions_90d"] >= 30) &
    ((df["engagement_rate"] < 30) | (df["scroll_rate"] < 30))
]
print("Low-engagement visible pages:", len(low_engagement))

# CTR by position tier — shows why position-adjustment matters
print(df.groupby("position_tier")["ctr"].mean().sort_index())
```
