## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/BrightTimothy/FlyRank-AI-Machine-Learning-Internship/main/data/raw/content_refresh_anonymized.csv")

lane4_cols = ["content_id", "impressions_90d", "clicks_90d", "ctr", "avg_position",
              "position_tier", "sessions_90d", "engagement_rate", "scroll_rate", "content_age_days"]
lane4_df = df[lane4_cols]
lane4_df.head()

expected_ctr = df.groupby("position_tier")["ctr"].transform("mean")
lane4_df["ctr_gap"] = df["ctr"] - expected_ctr

lane4_df[["content_id", "position_tier", "ctr", "ctr_gap"]].sort_values("ctr_gap").head()
```

## 5. Why ML beats a fixed rule here

A fixed rule (like ctr < 0.5) ignores that expected CTR varies nearly 10x across position tiers (0.15 in the deepest tier vs. 1.48 in top-3, from ML-02's real numbers) — so one flat threshold will over-flag deep pages and under-flag top-tier pages that are still genuinely underperforming for their tier. A learned scoring approach can weigh position, engagement, and volume signals together and adapt the threshold implicitly per tier, instead of me guessing one number that has to work everywhere at once.
